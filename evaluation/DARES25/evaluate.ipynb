```python
# init workflow
from agentic_tools.agents.evaluator_builder import ScoreCriteriaPerAnswerBuilder
from tqdm import tqdm
#generated_file = "../../gitlab/ows-eo-kg/evaluation/samples/big_eval_new.csv"
generated_file = "../../gitlab/ows-eo-kg/evaluation/samples/answers_w_structured_context.csv"

import pandas as pd
generated_df = pd.read_csv(generated_file)

def evaluate_answers_w_scores() -> pd.DataFrame:

    llm_full = "gpt-4.1-mini-2025-04-14"  # "mistral-medium"# "gpt-4o-mini-2024-07-18"


    all_res = []
    for i, row in tqdm(generated_df.iterrows(), total=len(generated_df)):
        answer_lst=[
            ("rag_answer", row["rag_answer"],  row["structured_context"]),
            ("zero_shot_answer", row["zero_shot_answer"], row["structured_context"]),
            ("rag_two_steps_answer", row["rag_two_steps_answer"], row["structured_context"])]


        validator_workflow = ScoreCriteriaPerAnswerBuilder(
            model_thread_id= "score_w_gpt4", 
            llm_name=llm_full,
            llm_num=1,
            experiment_name="score",
            temperature=0.7,
            answer_lst= answer_lst)
        
        q = row["question"]  # "Capital if France? "  #
        best = validator_workflow._invoke_graph(
            {"question": q}, id_=i)#.model_dump()  #["critical_questions"]
        
        
        rag_score = best['answer_scores_dict']["rag_answer"] 
        zero_shot_score = best['answer_scores_dict']["zero_shot_answer"] 
        rag_2score = best['answer_scores_dict']["rag_two_steps_answer"] 
        all_res.append({**row.to_dict(), "score_rag": rag_score,  "score_2rag": rag_2score, "score_zero_shot": zero_shot_score,
                        "score_is_rag_better": rag_score > zero_shot_score,
                        "score_is_2rag_better": rag_2score > zero_shot_score,
                        "score_instruction_template": best['instruction_template'],
                        "score_model" : llm_full,
                        **{f"score_2rag_criteria_{k}": v for k,v in best["answer_criteria_scores"]["rag_two_steps_answer"].items() },
                        **{f"score_rag_criteria_{k}": v for k,v in best["answer_criteria_scores"]["rag_answer"].items() },
                        **{f"score_0shot_criteria_{k}": v for k,v in best["answer_criteria_scores"]["zero_shot_answer"].items() }
                        
                        })
    def _apply_add_metadata(row):
        row["question_type"] = row["idx"].split("_")[-1]
        row["topic"] = "_".join(row["idx"].split("_")[:-1])
        return row

    scored_results = pd.DataFrame(all_res)
    scored_results = scored_results.apply(_apply_add_metadata, axis=1)
    return scored_results

scored_results = evaluate_answers_w_scores()
```

```python
# scored_results.to_csv("output/evaluation/scoring_structured_context_results.csv", index= False)


evals = {
    # "init": "output/evaluation/scoring_init_results.csv",
    "w_structured_context": "output/evaluation/scoring_structured_context_results.csv"
}
analysis_per_experiment = {}
criteria_lst = [
        "score_rag_criteria_factuality",
        "score_2rag_criteria_factuality",
        "score_0shot_criteria_factuality",
        "score_2rag_criteria_relevance",
        "score_rag_criteria_relevance",
        "score_0shot_criteria_relevance",
        "score_2rag_criteria_groundedness",
        "score_rag_criteria_groundedness",
        "score_0shot_criteria_groundedness",
        "score_2rag_criteria_helpfulness",
        "score_rag_criteria_helpfulness",
        "score_0shot_criteria_helpfulness",
        "score_2rag_criteria_depth",
        "score_rag_criteria_depth",
        "score_0shot_criteria_depth",
    ]




eval_df = pd.read_csv(evals["w_structured_context"])

eval_df = eval_df[
    [
        "idx",
        "model",
        "question_type",
        "topic",
        "score_rag",
        "score_zero_shot",
        "score_2rag",
        "score_is_rag_better",
        "score_is_2rag_better",
        "score_instruction_template",
        "score_model",
        "score_rag_criteria_factuality",
        "score_rag_criteria_relevance",
        "score_rag_criteria_groundedness",
        "score_rag_criteria_helpfulness",
        "score_rag_criteria_depth",
        "score_0shot_criteria_factuality",
        "score_0shot_criteria_relevance",
        "score_0shot_criteria_groundedness",
        "score_0shot_criteria_helpfulness",
        "score_0shot_criteria_depth",
        "score_2rag_criteria_factuality",
        "score_2rag_criteria_relevance",
        "score_2rag_criteria_groundedness",
        "score_2rag_criteria_helpfulness",
        "score_2rag_criteria_depth",
    ]
]
desc_stats = {}
# Overall
desc_stats["overall_score"] = (
    eval_df.groupby("model")[["score_2rag", "score_rag", "score_zero_shot"]]
    .mean()
    .reset_index()
)

#  Criteria


desc_stats["criteria"] = (
    eval_df.groupby("model")[criteria_lst].mean().reset_index()
)

desc_stats["question_type"] = (
    eval_df.groupby(["model", "question_type"])[
        ["score_2rag", "score_rag", "score_zero_shot"]
    ]
    .mean()
    .reset_index()
)

desc_stats["topic"] = (
    eval_df.groupby(["model", "topic"])[
        ["score_2rag", "score_rag", "score_zero_shot"]
    ]
    .mean()
    .reset_index()
)

# --- 5. : how often RAG is better? ---
desc_stats["rag_better_ratio_df"] = (
    eval_df.groupby("model")["score_is_rag_better"]
    .mean()
    .reset_index()
    .rename(columns={"score_is_rag_better": "rag_better_ratio"})
)

desc_stats["2rag_better_ratio_df"] = (
    eval_df.groupby("model")["score_is_2rag_better"]
    .mean()
    .reset_index()
    .rename(columns={"score_is_2rag_better": "rag_better_ratio"})
)
```

```python
##





def flatten_scores(df_scores):
    # Flatten the scores
    flatten_lst = []

    for _, row in df_scores.iterrows():
        for score_col in ["score_2rag", "score_rag", "score_zero_shot"]:
            exp = score_col.replace("score_", "")
            flat_elt = {
                "topic": row["topic"],
                "question_type": row["question_type"],
                "model": row["model"],
                "score": row[score_col],
                "experiment": exp
            }

            criteria_sublist = [c for c in criteria_lst if c.find(f'_{exp.replace("zero_shot", "0shot")}_') > -1]
            for criteria_col in criteria_sublist:
                flat_elt[f"score_{criteria_col.split('_')[-1]}"] = row[criteria_col]
            flatten_lst.append(flat_elt)
    return pd.DataFrame(flatten_lst)

# Define factors and their group columns

scores_flat_df = flatten_scores(eval_df)  # 
scores_flat_df.to_csv("output/evaluation/scores_flat.csv", index=False)
```

```python
import pandas as pd

df = scores_flat_df
# --- Scores to include ---
score_cols = ["score", "score_factuality", "score_relevance",
              "score_groundedness", "score_helpfulness", "score_depth"]

# --- Separate tables by model and aggregate mean ---
for model_name in df["model"].unique():
    df_model = (
        df[df["model"] == model_name]
        .groupby("experiment")[score_cols]
        .mean()
        .round(2)
    )

    latex_table = df_model.to_latex(index=True,
                                    caption=f"Mean scores per experiment for {model_name}",
                                    label=f"tab:{model_name.replace('-', '_')}",
                                    float_format="%.2f")
    print(f"% --- {model_name} ---")
    print(latex_table)
    print()
```
